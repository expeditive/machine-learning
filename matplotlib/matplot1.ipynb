#2D plot

use for bivariate analysis



categorical -> numerical and numerical -> numerical

use case (time series data)_

```python
#plotting a simple function
price = [2000,5000,8000,45000,1000,6700]
year = [2021,2022,2023,2024,2025,2026]

plt.plot(year,price)
```

```python
#data from a pd dataframe
batsman = pd.read_csv('/content/sharma-kohli.csv')
plt.plot(batsman['index'],batsman['V Kohli'])
```

```python
#plotting multiple plots
plt.plot(batsman['index'],batsman['V Kohli'])
plt.plot(batsman['index'],batsman['RG Sharma'])

plt.title("rohit vs virat in IPL")
plt.xlabel("season")
plt.ylabel("runs")
```

```python
#plotting multiple plots
plt.plot(batsman['index'],batsman['V Kohli'],linestyle='dotted',color='black',linewidth=1)
plt.plot(batsman['index'],batsman['RG Sharma'],linestyle='dashed',linewidth=2)

plt.title("rohit vs virat in IPL")
plt.xlabel("season")
plt.ylabel("runs")
```

```python
#plotting multiple plots
plt.plot(batsman['index'],batsman['V Kohli'],marker='D')
plt.plot(batsman['index'],batsman['RG Sharma'],marker='+',markersize=7)

plt.title("rohit vs virat in IPL")
plt.xlabel("season")
plt.ylabel("runs")
```

```python
#legend -> location
plt.plot(batsman['index'],batsman['V Kohli'],label='Virat')
plt.plot(batsman['index'],batsman['RG Sharma'],label='rohit')

plt.title("rohit vs virat in IPL")
plt.xlabel("season")
plt.ylabel("runs")

plt.legend()
```

```python
#limiting axes when there is one outlier which is too much from whole data

price = [2000,5000,8000,4500,1000,6700,3200000] #last price with high value
year = [2021,2022,2023,2024,2025,2026,2027]

plt.plot(year,price)
plt.ylim(0,50000)
```

```python
price = [2000,5000,8000,4500,1000,6700] #last price with high value
year = [2021,2022,2023,2024,2025,2026]

plt.plot(year,price)
plt.xlim(2022,2023)
```

```python
#plotting multiple plots with grids
plt.plot(batsman['index'],batsman['V Kohli'],marker='D')
plt.plot(batsman['index'],batsman['RG Sharma'],marker='+',markersize=7)

plt.title("rohit vs virat in IPL")
plt.xlabel("season")
plt.ylabel("runs")

plt.grid()
```

## BAR **chart**

```python
#simple bar chart
colors = ['black','blue','pink','purple']
childrens = [20,90,40,38]

plt.bar(colors,childrens)
```

```python
#horizontal barchart
plt.barh(colors,childrens)
```

```python
#color and label

bt = pd.read_csv('/content/batsman_season_record.csv')
bt
```

```python
# Sample data for plotting multiple bar graphs together
categories = ['A', 'B', 'C', 'D']
values1 = [3, 7, 2, 5]
values2 = [4, 6, 3, 4]

# Number of categories
n = len(categories)

# Position of bars on x-axis
x = np.arange(n)

# Width of each bar
width = 0.35

# Plotting bars side by side
plt.bar(x - width/2, values1, width, label='Dataset 1')
plt.bar(x + width/2, values2, width, label='Dataset 2')

# Labels and legend
plt.xlabel("Categories")
plt.ylabel("Values")
plt.title("Side-by-side Bar Graphs")
plt.xticks(x, categories)
plt.legend()
```

```python
#when the name turns very large
colors = ['black black black black','blue blue blue blue','pink pink pink pink','purple purple purple purple']
childrens = [20,90,40,38]

plt.bar(colors,childrens)
```

```python
#when the name turns very large
colors = ['black black black black','blue blue blue blue','pink pink pink pink','purple purple purple purple']
childrens = [20,90,40,38]

plt.bar(colors,childrens)
plt.xticks(rotation=45)
```

```python
#stacked bar chart
plt.bar(bt['batsman'],bt['2017'],label='2017')
plt.bar(bt['batsman'],bt['2016'],bottom=bt['2017'],label='2016')
plt.bar(bt['batsman'],bt['2015'],bottom=(bt['2016'])+(bt['2017']),label='2015')

plt.legend()
plt.show()
```
